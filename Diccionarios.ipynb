# Diferentes formas de crear diccionarios

```python
my_dict = {}
print(my_dict)
print(type(my_dict))
```

```python
my_dict = dict()
print(my_dict)
print(type(my_dict))
```

```python
# representando el diccionario como tal
my_dict = {
    1:"Uno",
    2:"Dos",
    3:"Tres",
    4:"Cuatro",
    5:"Cinco"
}
print(my_dict)
print(type(my_dict))
```

```python
# usando nombres de las llaves con valores
my_dict = dict(Uno=1, Dos=2, Tres=3, Cuatro=4, Cinco=5)
print(my_dict)
print(type(my_dict))
```

```python
# usando una lista de tuplas
my_dict = dict([
    (1, "Uno"),
    (2, "Dos"),
    (3, "Tres"),
    (4, "Cuatro"),
    (5, "Cinco")
])
print(my_dict)
print(type(my_dict))
```

## Método update

```python
my_dict = {
    1:"Uno",
    2:"Dos",
    3:"Tres",
    4:"Cuatro",
    5:"Cinco"
}
my_dict_2 = {
    6:"Seis",
    7:"Siete",
    8:"Ocho",
    9:"Nieve"
}

print(my_dict)
print(type(my_dict))

print(my_dict_2)
print(type(my_dict_2))
```

```python
my_dict.update(my_dict_2)
print(my_dict)
print(type(my_dict))
```

```python
my_dict = {
    1:"Uno",
    2:"Dos",
    3:"Tres",
    4:"Cuatro",
    5:"Cinco"
}
my_dict_2 = {
    1:"UNO",
    3:"TRES",
    4:"CUATRO"
}

print(my_dict)
print(type(my_dict))

print(my_dict_2)
print(type(my_dict_2))

my_dict.update(my_dict_2)
print(my_dict)
print(type(my_dict))
```

## Copy & Alias

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

alias = laptops # no es recomendable
copy = laptops.copy() # Siempre es recomendable

print(copy)
```

```python
# Usar un alias y hacer cambios en el alias, afecta al diccionario original
# En cambio, al usar copy, no afectas al diccionario original
```

```python
print(copy)
copy["HP"] = 9999.0
print(copy)
print(laptops)
```

## Métodos para manipulación de diccionarios

### clear
- Eliminar los elementos de un diccionario

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

print(laptops)
print(len(laptops))
```

```python
laptops.clear()
print(laptops)
print(len(laptops))
```

### copy
- Sirve para hacer una copia del diccionario

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

copy = laptops.copy()

print(laptops)
print(len(laptops))
print(copy)
print(len(copy))
```

### fromkeys
- Construir un nuevo diccionario tomando como referencias las keys en un diccionario

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

category = "Marca de Laptop"
new_dict = dict.fromkeys(laptops, category)
print(new_dict)
```

### get
- Obtener el valor de un diccionario de forma segura y con la opción de configurar un valor por default

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

print(laptops["Lenovo"]) # Forma insegura
print(laptops.get("Lenovo")) # Forma segura
```

```python
# Configurando valor por default en caso de que no esté presente
print(laptops.get("Asus", "No disponible"))
```

### items
- Obtener cada key y value en el diccionario de forma simultanea

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

for key, value in laptops.items():
    print("Para la llave '{}' el valor es {}".format(key, value))
```

### keys
- Obtener las keys de un diccionario

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

for key in laptops.keys():
    print("Una key del diccionario es: {}".format(key))
```

### values
- Obtener los values de un diccionario

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

for value in laptops.values():
    print("Un value del diccionario es: {}".format(value))
```

### pop
- Elimina la key y value de un diccionario pasando la key como parámetro
- La llave debe de estar presente en el diccionario, sino manda error

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

print(laptops)
print(len(laptops))

laptops.pop("Lenovo")
print(laptops)
print(len(laptops))
```

### popitem
- Elimina el índice -1 (El último elemento key-value)

```python
laptops = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

print(laptops)
print(len(laptops))

laptops.popitem()
print(laptops)
print(len(laptops))
laptops.popitem()
print(laptops)
print(len(laptops))
laptops.popitem()
print(laptops)
print(len(laptops))
```

### setdefault
- Configura un valor por default para cierta key en caso de que no esté presente en el diccionario
- Si la key está presente, no tendrá ningún efecto

```python
laptops_store_1 = {
    "HP" : 11999.0,
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

laptops_store_2 = {
    "Apple" : 24999.0,
    "Acer" : 12012.03,
    "Lenovo" : 17299.0,
    "Dell" : 15524.87
}

laptops_store_1.setdefault("HP", 10000.0)
laptops_store_2.setdefault("HP", 10000.0)

print(laptops_store_1.get("HP"))
print(laptops_store_2.get("HP"))
```
