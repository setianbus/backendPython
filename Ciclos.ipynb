# Ciclos
- También llamados Declaraciones Iterativas
- Son usados para repetir un bloque específico de código
- Si vas a hacer un proceso una y otra vez, entonces puedes usar un ciclo para ello

- Tipos
    - for
    - while

## Ciclo For
- Son usados para iterar los valores "n" número de veces
- Podemos usar la palabra reservada "for"
- Puede iterar los valores de estructuras de datos como listas, tuplas, sets o diccionarios
- También puede iterar sobre cada letra de una cadena
- Itera en una secuencia en orden mientras ejecuta un bloque de código

```python
my_str = "Backend Python"
print(my_str)
print(type(my_str))

for letter in my_str:
    print(letter)
```

```python
my_str = "Backend Python"
print(my_str)
print(type(my_str))

for letter in my_str:
    print("antes")
    print(letter)
    print("después")
```

```python
my_str = "Backend Python"
print(my_str)
print(type(my_str))

for letter in my_str:
    print("Ciclo for iterando con la letra: ", letter)
```

```python
my_list = [1,2,3,4,5]
print(my_list)
print(type(my_list))

for number in my_list:
    print("Ciclo for iterando con el número: ", number)
```

```python
for number in range(1,6):
    print("Ciclo for iterando con el número: ", number)
```

## Números pares e impares con un ciclo for

```python
# Números Pares
for number in range(1,21):
    if number%2==0:
        print(number, "es un número par")
```

```python
# Números nones
for number in range(1,21):
    if number%2==1:
        print(number, "es un número non")
```

## Ciclo FOR anidado
- Un ciclo for dentro de otro ciclo for

```python
nouns = ["balón", "bote", "celular", "libro"]
adjectives = ["pequeño", "grande", "redondo", "cuadrado", "rojo", "azul"]

for noun in nouns:
    for adjective in adjectives:
        print("{} {}".format(noun,adjective))
```

## Ciclo While
- Solo se ejecuta si la condición es verdadera
- Se puede usar para correr un proceso de manera perpetua
- Consiste de una condición y un bloque de código
- Valida la condición antes de que el bloque sea ejecutado

```python
my_number = 0
print("Previo al ciclo while")
while my_number < 5:
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1
print("Fin del ejemplo")
```

```python
my_number = 0
print("Previo al ciclo while")
while my_number >=0 and my_number < 5:
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1
else:
    print(my_number, "Es un número negativo o mayor a 4")
print("Fin del ejemplo")
```

```python
my_number = -3
print("Previo al ciclo while")
while my_number >=0 and my_number < 5:
    print("El valor de my_number es: ", my_number)
    my_number += 1 # my_number = my_number + 1
else:
    print(my_number, "Es un número negativo o mayor a 4")
print("Fin del ejemplo")
```

## Iterando en estructuras de datos

### Listas

```python
emails = ["sebas@hotmail.com","chore@hotmail.com","sebastian@hotmail.com"]

for email in emails:
    print(email)
```

```python
user_ids = [2634, 9274, 1937]

for user_id in user_ids:
    print(user_id)
```

### Diccionarios

```python
user_ids = [2634, 9274, 1937]
emails = ["sebas@hotmail.com","chore@hotmail.com","sebastian@hotmail.com"]
my_dict = dict(zip(user_ids, emails))

print(my_dict)

for user_id, email in my_dict.items():
    print("El user {} tiene el correo {}".format(user_id, email))
```

## Declaraciones especiales
- break: detiene el ciclo
- continue: se salta la iteración actual (sin ejecutar el código posterior) y pasar a la siguiente iteración
- pass: se usa para escribir ciclos vacíos o para escribir funciones o clases vacias

### break
- Es usado para detener un ciclo basado en ciertas condiciones
- Te ayuda a reducir el número de iteraciones innecesarias

```python
my_number = 0
past_number = None
print("Previo al ciclo while")
while my_number < 5:
    if past_number == my_number:
        break
    print("El valor de my_number es: ", my_number)
    past_number = my_number
print("Fin del ejemplo")
```

```python
my_number = 0
while my_number < 5:
    print("El valor de my_number es: ", my_number)
    my_number += 1
print("Fin del ejemplo")
```

```python
my_number = 0
while my_number < 5:
    if my_number == 3:
        break
    print("El valor de my_number es: ", my_number)
    my_number += 1
print("Fin del ejemplo")
```

## continue
- Salta la iteración actual y pasa a la siguiente ietración
- Se puede usar tanto en ciclos for como en ciclos while

```python
my_number = 0
while my_number < 5:
    if my_number == 3:
        my_number +=1
        print("Antes del continue")
        continue
        print("después del continue")
    print("El valor de my_number es: ", my_number)
    my_number += 1
print("Fin del ejemplo")
```

```python
for number in range(1,21):
    if number == 10:
        continue
    print("Ciclo for iterando con el número: ", number)
```
