# Funciones en Python
- Es un conjunto de operaciones
- Puede ser usada como un molde reutilizable
- Puede regresar un valor
- Puede ahorrar mucho tiempo de desarrollo
- Puede aceptar parámetros para mandar valores al programa/función

## Funciones definidas por el usuario
- Las funciones son comunmente usadas en todos los programas
- Código reutilizable para ejecutar tareas específicas
- Pueden regresar valores
- Las funciones actúan como un molde para tareas repetidas
- Las funciones pueden ser creadas con la palabra clave "def"
- Podemos usar la palabra clave "return" para regresar un valor

Notas:
- La palabra clave "def" es obligatoria para definir una función
- Los parámetros/parameters son opcionales
- Las instrucciones dentro de la función son obligatorias, en caso de no específicar estas instrucciones se debe sustituir con un "pass"
- La palabra clave "return" es opcional en una función

```python
def my_function (parameters):
    print("Paso 1")
    print("Paso 2")
    print("Paso 3")
    return "Completado"
```

## Parámetros de Funciones
- Los parámetros son opcionales en una función
- Podemos declarar los parámetros en la definición de la función
- También podemos llamarles argumentos
- Los parámetros son usados para pasar valores dinámicamente a la función
- Tenemos 2 tipos de parámetros:
    - Formales
    - Reales

## Tipos de parámetros

Parámetros Formales
- Son los valores usados en la creación/definición de la función

Parámetros Reales
- Son los valores usados al llamar/invocar la función

```python
def print_number(number): # Parámetros Formales
    print("El valor de number es: ", number)
```

```python
print_number(3800) #Parámetro Real
```

## Invocar Funciones

```python
def plus_ten(number):
    return number + 10

def minus_ten(number):
    return number - 10

def another_function(f, number):
    result = f(number)
    return result
```

```python
plus_ten(200)
```

```python
minus_ten(200)
```

```python
another_function(plus_ten, 200)
```

## Funciones Anidadas
- Es una función dentro de otra función
- Así como hay listas anidadas o ciclos anidados
- Las funciones internas pueden ser útiles al ejecutar tareas complejas más de una vez dentro de una función
- Pueden ayudar a evitar ciclos o código duplicado

Usos:
- Simplificar la lógica y estructura
- Construir nuevas funciones basadas en parámetros
- Pasar funciones a otras funciones para generar resultados precisos

```python
def my_function():
    def nested_function():
        print("Función Anidada")
    return nested_function
```

```python
result = my_function()
```

```python
result()
```

```python
nested_function()  #No se puede llamar directamente a una función dentro de otra función
```

```python
# Caso un poco mas realista
```

```python
def logger(func):
    def log_func(*args):
        print("Corriendo '{}' con los parámetros {}".format(func.__name__, args))
        print(func(*args))
    return log_func

def add(x, y):
    return x + y

def sub(x, y):
    return x - y

add_logger = logger(add)
sub_logger = logger(sub)
```

```python
add(3,2)
```

```python
add_logger(3,2)
```

```python
sub_logger(3,2)
```

## Función enumerate

```python
programing_lenguages = ["Python", "Go", "Javascript", "Java", "Scala"]

for language_name in programing_lenguages:
    print(language_name)
```

```python
programing_lenguages = ["Python", "Go", "Javascript", "Java", "Scala"]

for i, language_name in enumerate(programing_lenguages):
    print("El lenguaje en el índice {} es {}".format(i, language_name))
```

```python
for i, language_name in enumerate(programing_lenguages):
    print("El lenguaje en el índice {} es {}".format(i+1, language_name))
```

```python
for i, language_name in enumerate(programing_lenguages, start=1):
    print("El lenguaje en el índice {} es {}".format(i, language_name))
```

## Funciones lambda 

- Es una función anónima (sin nombre)
- Puede recibir cualquier número de argumentos pero solo puede tener una expresión
- "lambda" es una palabra clave en Python que representa una función anónima dentro de otra función
- La ventaja principal es usarla en combinación con built-in functions como map o filter
- La función es regresada como un valor para otra función

```python
lambda x:x+10 #Por cada entrada "x" va a regresar esa misma "x" + 10
```

```python
# Se le puede asignar un valor a la función lambda
plus_ten = lambda x:x+10
```

```python
print(plus_ten(10))
print(plus_ten(50))
print(plus_ten(140))
```

```python
my_list = [10, 50, 140, 200, 700, 16, 18]
result = map(plus_ten, my_list) # map aplica una función a cada elemento de la lista
```

```python
result
```
