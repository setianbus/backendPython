## Creando clases

```python
class Car:   # Molde

    def __init__(self, mark, model, year): # constructor de la clase
        self.mark = mark
        self.model = model     # asignamos las instancias a si mismo
        self.year = year
        print("Una instancia de la clase se ha creado exitosamente")

    def print_attributes(self):
        print("Marca: ", self.mark)
        print("Modelo: ", self.model)
        print("Año: ", self.year)
```

## Creando Objetos

```python
car1 = Car("Tesla", "Model 3", "2020")
```

```python
car1.print_attributes()
```

## Variables de clase

```python
class Car:   # Molde
    opus = 0 
    
    def __init__(self, mark, model, year): # constructor de la clase
        Car.opus += 1
        self.mark = mark
        self.model = model     # asignamos las instancias a si mismo
        self.year = year
        print("Una instancia de la clase se ha creado exitosamente")

    def print_attributes(self):
        print("Marca: ", self.mark)
        print("Modelo: ", self.model)
        print("Año: ", self.year)
```

```python
car1 = Car("Tesla", "Model 3", "2020")
car2 = Car("Tesla", "Model Y", "2019")
```

## Garbage collection

```python
class Car:   # Molde
    total = 0 
    
    def __init__(self, mark, model, year): # constructor de la clase
        Car.total += 1
        self.mark = mark
        self.model = model     # asignamos las instancias a si mismo
        self.year = year
        print("Una instancia de la clase se ha creado exitosamente")

    def __del__(self):
        Car.total -= 1
        
    def print_attributes(self):
        print("Marca: ", self.mark)
        print("Modelo: ", self.model)
        print("Año: ", self.year)
```

```python
car1 = Car("Tesla", "Model 3", "2020")
car2 = Car("Tesla", "Model Y", "2019")
car3 = Car("Tesla", "Model X", "2021")
car4 = Car("Tesla", "Model S", "2022")
```

```python
print("Hay ", Car.total, " carros creados")
```

```python
del car4
```

```python
print("Hay ", Car.total, " carros creados")
```

## Herencia

```python
class Car:  # Clase "Papá"
    super_attribute = "Atributo de la súper clase / Clase padre"
    def __init__(self):
        print("Iniciación de la súper clase / Clase padre")
    def super_method(self):
        print("Método de la súper clase / Clase padre")

class CollectionCar(Car):  # Clase hija que va a heredar
    sub_attribute = "Atributo de la sub clase / Clase hija"
    def __init__(self):
        print("Iniciación de la sub clase / Clase hija")
    def sub_method(self):
        print("Método de la sub clase / Clase hija")
```

```python
sub_car = CollectionCar()
```

```python
sub_car.super_method()
sub_car.sub_method()
print(sub_car.sub_attribute)
print(sub_car.super_attribute)
```

## Usando una SubClase como SuperClase

```python
class Car:  # Clase "Papá"
    super_attribute = "Atributo de la súper clase / Clase padre"
    def __init__(self):
        print("Iniciación de la súper clase / Clase padre")
    def super_method(self):
        print("Método de la súper clase / Clase padre")

class CollectionCar(Car):  # Clase hija que va a heredar
    sub_attribute = "Atributo de la sub clase / Clase hija"
    def __init__(self):
        print("Iniciación de la sub clase / Clase hija")
    def sub_method(self):
        print("Método de la sub clase / Clase hija")

class SubCollectionCar(CollectionCar):  # Clase nieta que va a heredar
    sub_sub_attribute = "Atributo de la sub clase / Clase nieta"
    def __init__(self):
        print("Iniciación de la sub clase / Clase nieta")
    def sub_sub_method(self):
        print("Método de la sub clase / Clase nieta")
```

```python
sub_sub_car = SubCollectionCar()
```

```python
sub_sub_car.super_method()
sub_sub_car.sub_method()
sub_sub_car.sub_sub_method()
print(sub_sub_car.sub_attribute)
print(sub_sub_car.super_attribute)
print(sub_sub_car.sub_sub_attribute)
```
