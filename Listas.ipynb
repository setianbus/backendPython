## Diferentes formas de crear listas

```python
my_list = [3]*5
print(my_list)
```

```python
my_list = [1,2,3,4,5]
print(my_list)
```

```python
my_list = []
for number in range(10):
    my_list.append(number)
print(my_list)
```

```python
my_list = []
for number in range(10):
    my_list+=[number]
print(my_list)
```

```python
my_list = [number for number in range(15)]
print(my_list)
```

## Crear listas usando range y numpy

```python
#range(inicio, fin, paso)
my_list = [range(1,5,1)]   #Lista de rangos con un solo elemento
print(my_list)
```

```python
#range(inicio, fin, paso)
my_list = [*range(1,5,1)]   #Lista de números
print(my_list)
```

```python
#range(inicio, fin, paso)
my_list = [*range(1,20,4)]  
print(my_list)
```

```python
import numpy as np
my_list = np.arange(1,20,4).tolist()
print(my_list)
```

## Ordenando Listas

```python
my_list = [5,3,1,15,2,20]
print(my_list)
my_list.sort() #Ordena la lista en la misma variable, por lo que se pierde la lista original desordenada
print(my_list)
```

```python
my_list = [5,3,1,15,2,20]
print(my_list)
my_list.sort(reverse=True) 
print(my_list)
```

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
my_list.sort() #Ordena alfabeticamente
print(my_list)
```

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
my_list.sort(reverse=True) #Ordena alfabeticamente
print(my_list)
```

## Conversión de listas

### Cadenas a Listas

```python
# Separar palabras
my_str = "Curso de Backend Python"
my_list = my_str.split()
print(my_str)
print(my_list)
```

```python
# Separar caracteres
my_str = "Curso de Backend Python"
my_list = list(my_str)
print(my_str)
print(my_list)
```

```python
# Separar caracteres en una lista de listas
my_str = "Curso de Backend Python"
my_list = list(map(list, my_str))
print(my_str)
print(my_list)
```

### Listas a Cadenas

```python
# Usando espacios entre palabras
my_list = ['Curso', 'de', 'Backend', 'Python']
my_str = " ".join(my_list)
print(my_list)
print(my_str)
```

```python
my_list = ['C', 'u', 'r', 's', 'o', ' ', 'd', 'e', ' ', 'B', 'a', 'c', 'k', 'e', 'n', 'd', ' ', 'P', 'y', 't', 'h', 'o', 'n']
my_str = "".join(my_list)
print(my_list)
print(my_str)
```

```python
my_list = [['C'], ['u'], ['r'], ['s'], ['o'], [' '], ['d'], ['e'], [' '], ['B'], ['a'], ['c'], ['k'], ['e'], ['n'], ['d'], [' '], ['P'], ['y'], ['t'], ['h'], ['o'], ['n']]
my_str = "".join(["".join(element) for element in my_list])
print(my_list)
print(my_str)
```

# Métodos para manipulación de listas

## append
- Agregar un elemento al final de una lista

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.append("delfin")
print(my_list)
print(len(my_list))
```

```python
my_list.append("gato")
print(my_list)
print(len(my_list))
```

```python
my_list = [300,500,100,200]
print(my_list)
print(len(my_list))
```

```python
my_list.append(111)
print(my_list)
print(len(my_list))
```

```python
my_list.append(473)
print(my_list)
print(len(my_list))
```

## insert
- Agregar un valor a una lista en una posición específica

- insert(índice, valor)

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.insert(2, "delfin")
print(my_list)
print(len(my_list))
```

```python
my_list.insert(1, "gato")
print(my_list)
print(len(my_list))
```

## clear
- Eliminar todos los elementos de la lista

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.clear()
print(my_list)
print(len(my_list))
```

## copy
- Copiar los elementos de una lista

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list_copy = my_list.copy()
print(my_list_copy)
print(len(my_list_copy))
```

```python
my_list_copy.clear()
print(my_list_copy)
print(len(my_list_copy))
print(my_list)
print(len(my_list)) # No afecta a la lista original
```

## count
- Cuenta cuantas veces aparece un elemento en una lista

```python
my_list = [100, 300, 500, 100, 200, 100, 300]
print(my_list)
```

```python
print("El número 100 aparece {} veces en la lista".format(my_list.count(100)))
```

```python
print("El número 100 aparece {} veces en la lista".format(my_list.count(300)))
```

## index
- Obtiene el primer indice de una lista que hace match con el elemento solicitado

```python
my_list = [100, 300, 500, 100, 200, 100, 300]
print(my_list)
```

```python
print("El índice de la primera aparición del número 100 es: {}".format(my_list.index(100)))
```

```python
print("El índice de la primera aparición del número 100 es: {}".format(my_list.index(500)))
```

## extend

-Combinar dos listas

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list_2 = ["delfín", "gato"]
print(my_list_2)
print(len(my_list_2))
```

```python
my_list.extend(my_list_2)
print(my_list)
print(len(my_list))
```

## pop
- Eliminar elementos de una lista basado en los índices

```python
my_list = [100,300,500,700,300,100,200]
print(my_list)
print(len(my_list))
```

```python
# Elimina índice -1 por default
my_list.pop()
print(my_list)
print(len(my_list))
```

```python
my_list.pop(2)
print(my_list)
print(len(my_list))
```

## remove
- Elimina la primera aparición de un valor en una lista

```python
my_list = [100,300,500,700,300,100,200]
print(my_list)
print(len(my_list))
```

```python
my_list.remove(100)
print(my_list)
print(len(my_list))
```

```python
my_list.remove(500)
print(my_list)
print(len(my_list))
```

```python
# Error - el valor debe de estar en la lista
my_list.remove(1000)
print(my_list)
print(len(my_list))
```

## reverse
- Invierte el orden de una lista

```python
my_list = ["caballo", "abeja", "jirafa", "ballena"]
print(my_list)
print(len(my_list))
```

```python
my_list.reverse()
print(my_list)
print(len(my_list))
```

## sort
- Ordena los elementos de una lista

```python
my_list = [100,300,500,700,300,100,200]
print(my_list)
my_list.sort()
print(my_list)
```
